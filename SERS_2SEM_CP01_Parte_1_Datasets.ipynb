7.	Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo critério.

```python
print(f'O dataset df_alto_consumo possui {df1_altoConsumo.shape[0]} linhas e {df1_altoConsumo.shape[1]} colunas')
print(f'O dataset df_alto_consumo_t1 possui {df1_altoConsumo_t1.shape[0]} linhas e {df1_altoConsumo_t1.shape[1]} colunas')
```

Ao incluir a temperatura acima da média de T1 como segundo critério, o número de registros no DataFrame `df1_altoConsumo_t1` diminui em comparação com `df1_altoConsumo`. Isso ocorre porque estamos aplicando uma condição mais restritiva, filtrando apenas os casos de alto consumo que também ocorrem quando a temperatura no setor 1 (T1) está acima de sua média.

Isso demonstra que nem todos os eventos de alto consumo estão associados a temperaturas elevadas no Setor 1. A inclusão desse critério nos permite focar em um subconjunto de eventos de alto consumo que compartilham uma característica ambiental específica, o que pode ser útil para entender padrões de consumo em diferentes condições.

Uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio. O dataset é o mesmo utilizado como referência em aula, mas o critério de análise será diferente.

1. Carregue a nova amostra e organize os atributos utilizando nomes mais simples.

```python
df6 = pd.read_csv('/content/household_power_consumption_SAMPLE_01.csv')

df6.columns
```

```python
df6.rename(columns={'Date':'Data',
                    'Global_active_power':'Potencia_Ativa',
                    'Global_reactive_power':'Potencia_Reativa',
                    'Voltage':'Voltagem',
                    'Global_intensity':'Corrente',
                    'Sub_metering_1':'Sub_Medicao_1',
                    'Sub_metering_1':'Sub_Medicao_2',
                    'Sub_metering_1':'Sub_Medicao_3'}, inplace=True)
```

2. Determine o valor máximo de potência ativa.

```python
maxPA = df6['Potencia_Ativa'].max()

maxPA
```

3. Calcule 75% do valor máximo e crie um DataFrame contendo registros acima desse limite.

```python
limiarPA = maxPA * 0.75

limiarPA
```

```python
potenciaAtiva_df6 = df6[df6['Potencia_Ativa'] > limiarPA]

potenciaAtiva_df6.head()
```

4. Calcule a quantidade e o percentual de registros selecionados.

```python
totalRegistros_df6 = len(df6)

nRegistrosPA = len(potenciaAtiva_df6)

percentualPA_df6 = (nRegistrosPA / totalRegistros_df6) * 100

print(f"Número de registros com consumo acima do limiar: {nRegistrosPA}")
print(f"Total de registros da amostra: {totalRegistros_df6}")
print(f"Percentual da amostra que representa: {percentualPA_df6:.2f}%")
```

5. Calcule a corrente média da amostra.

```python
correnteMedia = df6['Corrente'].mean()

correnteMedia
```

6. Crie um segundo DataFrame contendo simultaneamente potência ativa acima de 75% do máximo e corrente acima da média.

```python
potenciaAtiva_correnteMedia_df6 = potenciaAtiva_df6[potenciaAtiva_df6['Corrente'] > correnteMedia]

len(potenciaAtiva_correnteMedia_df6)
```

7. Compare os dois conjuntos e explique o efeito da inclusão da corrente como segunda condição.

```python
print(f'O DataFrame potenciaAtiva_df6 possui {len(potenciaAtiva_df6)} registros.')
print(f'O DataFrame potenciaAtiva_correnteMedia_df6 possui {len(potenciaAtiva_correnteMedia_df6)} registros.')
```

Ao incluir a condição de `Corrente` acima da média da amostra, o número de registros no DataFrame `potenciaAtiva_correnteMedia_df6` pode diminuir em relação a `potenciaAtiva_df6`. No entanto, como no resultado anterior ambos os DataFrames tiveram o mesmo número de registros (58), isso indica que todos os registros com `Potencia_Ativa` acima de 75% do máximo também possuem `Corrente` acima da média da amostra. Em outras palavras, para este recorte de dados, alta potência ativa está sempre associada a uma corrente acima da média. Isso pode sugerir que a `Corrente` é um fator diretamente proporcional à `Potencia_Ativa` nestes cenários de alta demanda, e que a condição de 'corrente acima da média' não impôs uma restrição adicional que eliminasse registros do conjunto de alta potência ativa.
